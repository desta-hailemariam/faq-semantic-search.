```python
## 2. Data Cleaning and Validation
```

```python
df.isnull().sum()
print("Duplicate questions:")
print(df.duplicated(subset=["question"]).sum())
```

```python

# delete unneccessary space
df["question"] = df["question"].str.strip()

df["answer"] = df["answer"].str.strip()
display(df.head())
# remove rows having missing values
df = df.dropna(subset=["question", "answer"])
```

```python
# remove dublicate rows and columens
df = df.drop_duplicates(
    subset=["question"],
    keep="first"
)
print("Remaining duplicate questions:",
      df.duplicated(subset=["question"]).sum())
```

```python
df.to_csv("data/faqs_cleaned.csv", index=False)
```

```python
## 3. Text Chunking
```

```python
df = pd.read_csv("data/faqs_cleaned.csv")

display(df.head())
```

```python
chunks = []

for index, row in df.iterrows():
    text = f"Question: {row['question']}\nAnswer: {row['answer']}"
    chunks.append(text)
```

```python
print("Total chunks:", len(chunks))

print("\nFirst chunk:")
print(chunks[0])
```

```python
4.generste embeddings
```

```python
!pip install sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = model.encode(chunks)

print("Number of embeddings:", len(embeddings))
print("Embedding dimensions:", embeddings.shape)
```
